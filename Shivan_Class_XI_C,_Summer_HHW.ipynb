Program 1: Unit Converter (Celsius $\leftrightarrow$ Fahrenheit)

```python
choice = input("Enter '1' for C to F or '2' for F to C: ")

if choice == '1':
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = (celsius * 9/5) + 32
    print(f"{celsius}°C is equal to {fahrenheit}°F")
elif choice == '2':
    fahrenheit = float(input("Enter temperature in Fahrenheit: "))
    celsius = (fahrenheit - 32) * 5/9
    print(f"{fahrenheit}°F is equal to {celsius}°C")
else:
    print("Invalid selection.")
```

Program 2: Triangle Validity Check

```python
side1 = float(input("Enter length of first side: "))
side2 = float(input("Enter length of second side: "))
side3 = float(input("Enter length of third side: "))

if (side1 + side2 > side3) and (side2 + side3 > side1) and (side1 + side3 > side2):
    print(f"A triangle can be formed with sides {side1}, {side2}, and {side3}.")
else:
    print("A triangle cannot be formed. Sum of any two sides must be greater than the third.")
```

Program 3: Circle Metrics (Area & Circumference)

```python
radius = float(input("Enter the radius of the circle: "))

pi = 3.141592653589793
area = pi * (radius ** 2)
circumference = 2 * pi * radius

print(f"Radius: {radius}")
print(f"Area: {area:.2f}")
print(f"Circumference: {circumference:.2f}")
```

Program 6: Quadratic Equation Root Finder

```python
a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

discriminant = (b ** 2) - (4 * a * c)

if discriminant > 0:
    root1 = (-b + (discriminant ** 0.5)) / (2 * a)
    root2 = (-b - (discriminant ** 0.5)) / (2 * a)
    print(f"Two distinct real roots: {root1:.2f} and {root2:.2f}")
elif discriminant == 0:
    root = -b / (2 * a)
    print(f"One repeated real root: {root:.2f}")
else:
    real_part = -b / (2 * a)
    imaginary_part = (abs(discriminant) ** 0.5) / (2 * a)
    print(f"Complex roots: {real_part:.2f} + {imaginary_part:.2f}i and {real_part:.2f} - {imaginary_part:.2f}i")
```

Program 11: Factorial Calculation (using while loop)

```python
num = int(input("Enter a positive integer: "))

if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1
    original_num = num
    while num > 1:
        factorial *= num
        num -= 1
    print(f"The factorial of {original_num} is {factorial}")
```

Program 13: Fibonacci Series Sequence Generator

```python
n = int(input("Enter the number of terms for the Fibonacci sequence: "))

if n <= 0:
    print("Please enter a positive integer greater than zero.")
elif n == 1:
    print("Fibonacci sequence: 0")
else:
    t1, t2 = 0, 1
    print("Fibonacci sequence:")
    print(t1, end=" ")
    print(t2, end=" ")
    count = 2
    while count < n:
        nth = t1 + t2
        print(nth, end=" ")
        t1 = t2
        t2 = nth
        count += 1
    print()  # For line break
```

Program 14: Prime or Composite Number Checker

```python
num = int(input("Enter a positive integer to check: "))

if num <= 1:
    print(f"{num} is neither prime nor composite.")
else:
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{num} is a Prime number.")
    else:
        print(f"{num} is a Composite number.")
```

Program 15: Armstrong Number Verification

```python
num = int(input("Enter a 3-digit number: "))

if 100 <= num <= 999:
    temp = num
    total_sum = 0
    while temp > 0:
        digit = temp % 10
        total_sum += digit ** 3
        temp //= 10

    if num == total_sum:
        print(f"{num} is an Armstrong number.")
    else:
        print(f"{num} is not an Armstrong number.")
else:
    print("The entered number is not a valid 3-digit number.")
```

Program 16: Sum of Series Calculation ($1 + x + x^2 + ... + x^n$)

Program 19: Greatest Common Divisor (GCD/HCF) Finder

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

a, b = abs(num1), abs(num2)
while b > 0:
    a, b = b, a % b

print(f"The Greatest Common Divisor (GCD) of {num1} and {num2} is {a}")
```

Program 20: Least Common Multiple (LCM) Calculator

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

a, b = abs(num1), abs(num2)
while b > 0:
    a, b = b, a % b

gcd = a
lcm = (abs(num1) * abs(num2)) // gcd

print(f"The Least Common Multiple (LCM) of {num1} and {num2} is {lcm}")
```

Program 21: Decimal to Binary Equivalent Converter

```python
decimal_num = int(input("Enter a decimal number: "))

if decimal_num == 0:
    binary_str = "0"
else:
    binary_str = ""
    temp = abs(decimal_num)
    while temp > 0:
        remainder = temp % 2
        binary_str = str(remainder) + binary_str
        temp //= 2

    if decimal_num < 0:
        binary_str = "-" + binary_str

print(f"The binary equivalent of {decimal_num} is {binary_str}")
```
